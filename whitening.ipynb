```python
import os
import time
import json
import random
from pathlib import Path
from typing import Callable
from dotenv import load_dotenv
import numpy as np
from tqdm.notebook import tqdm
import pandas as pd
from torch import amp
from transformers import AutoTokenizer, LlamaForCausalLM, LlamaModel, logging
import transformers
from datasets import load_dataset, concatenate_datasets

logging.set_verbosity_error()

# Set the seed for reproducability
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
transformers.set_seed(SEED)

# BATCH SIZE
batch_size = 4

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
model_hf_mappings = {
    "llama-base": "meta-llama/Meta-Llama-3-8B",
    "llama-instruct": "meta-llama/Meta-Llama-3-8B-Instruct",
    "categorical-refusal": "tomg-group-umd/zephyr-llama3-8b-sft-refusal-n-contrast-multiple-tokens",
    "binary-refusal": "tomg-group-umd/zephyr-llama3-8b-sft-refusal-n-contrast-single-token",
}
```

```python
start_time = time.time()

model_name = "categorical-refusal"

assert (
    model_name in model_hf_mappings.keys()
), "The specified model name is not supported. Please choose a valid model"

model_id = model_hf_mappings[model_name]

tokenizer = AutoTokenizer.from_pretrained(model_id)

model = LlamaForCausalLM.from_pretrained(
    model_id,
    trust_remote_code=True,
    device_map="auto",
    output_hidden_states=True,
    torch_dtype=torch.float16,
    low_cpu_mem_usage=True,
)
model.eval()
model.to(device)

end_time = time.time()
print(f"Model download time: {(end_time - start_time):.4f} seconds")
```

```python
from datasets import load_dataset
from torch.utils.data import DataLoader
import os

# ds = load_dataset("allenai/coconot", "contrast")
# ds["test"].set_format(type="python")  # keep as dict of lists of strings

# loader = DataLoader(
#     ds["test"],
#     batch_size=4,
#     shuffle=True,
#     num_workers=0,
#     pin_memory=False,  # strings can't be pinned in memory
# )

# for batch in loader:
#     print(batch)
#     break

# --- Load dataset ---
dataset = load_dataset("allenai/coconot", "contrast")
prompts = dataset["test"]["prompt"]  # list of strings

# --- Create DataLoader for batching ---
class PromptDataset(torch.utils.data.Dataset):
    def __init__(self, prompts):
        self.prompts = prompts

    def __len__(self):
        return len(self.prompts)

    def __getitem__(self, idx):
        return self.prompts[idx]

def collate_fn(batch):
    return tokenizer(
        batch,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=512,
    )

dataloader = DataLoader(
    PromptDataset(prompts),
    batch_size=batch_size,
    shuffle=False,
    collate_fn=collate_fn,
)

# --- Function to get layer 16 last-token activations ---
def get_layer_16_last_token_activations(inputs):
    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    with torch.no_grad():
        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        hidden_states = outputs.hidden_states  # tuple of (layer_count + 1)
        layer_16 = hidden_states[16]  # (batch_size, seq_len, hidden_size)

        # Find lengths of each sequence to index the last non-padding token
        lengths = attention_mask.sum(dim=1) - 1  # subtract 1 to get last token index
        last_token_activations = layer_16[torch.arange(layer_16.size(0)), lengths]

    return last_token_activations.cpu()

# --- Loop through all prompts and extract activations ---
all_activations = []

for batch in dataloader:
    batch_activations = get_layer_16_last_token_activations(batch)
    all_activations.append(batch_activations)

# --- Concatenate all activations ---
all_activations = torch.cat(all_activations, dim=0)  # shape: (num_samples, hidden_size)

print("Final activation shape:", all_activations.shape)
```

## Compute the Covariance Matrix: Sigma

```python
A = all_activations.float()  # Make sure dtype is float for stability
A_centered = A - A.mean(dim=0, keepdim=True)
# Covariance: [4096, 4096]
cov = (A_centered.T @ A_centered) / A_centered.shape[0]
cov.shape
```

## Compute Whitening Matrix (Inverse Square Root) using SVD

```python
eps = 1e-4
U, S, V = torch.linalg.svd(cov + eps * torch.eye(cov.size(0)))
Sinv_sqrt = torch.diag(S.clamp_min(eps).rsqrt())
whitening_matrix = U @ Sinv_sqrt @ U.T  # [4096, 4096]
whitening_matrix
```

## Apply Whitening Matrix to Steering Vectors
